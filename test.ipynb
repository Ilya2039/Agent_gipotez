```python
# Установи, если не стоят:
# !pip install -U langchain-community gigachat python-dotenv

import os
from dotenv import load_dotenv
from langchain_community.chat_models.gigachat import GigaChat
from langchain_core.messages import HumanMessage

# Загружаем токен
load_dotenv()
os.environ["GIGACHAT_AUTH"] = os.getenv("GIGACHAT_AUTH", "").strip() or "ВСТАВЬ_СВОЙ_ТОКЕН_СЮДА"

token = os.environ["GIGACHAT_AUTH"]
if not token:
    raise RuntimeError("Не найден токен GIGACHAT_AUTH!")

# ⚙️ Инициализация GigaChat с reasoning-моделью
llm = GigaChat(
    credentials=token,
    scope="GIGACHAT_API_CORP",   # или GIGACHAT_API_PERS
    model="GigaChat-2-Reasoning", # 🚀 нужная тебе модель
    verify_ssl_certs=False,
    temperature=0.3
)

# 💬 Формируем сообщение и делаем запрос
message = [HumanMessage(content="Объясни, почему небо кажется голубым?")]
response = llm.invoke(message)

print("Ответ модели:")
print(response.content)
```
